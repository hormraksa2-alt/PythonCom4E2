## Exercise 2: Iterator protocol deep dive

- Create iterator from a string and consume values with `next()` until exhausted.
- Handle `StopIteration` explicitly.
- Print total number of consumed characters.

```python
# Write your solution here
text = "Python"
text_iter = iter(text)

count = 0
print("Consuming iterator values:")
try:
    while True:
        char = next(text_iter)
        print(f"Read character: {char}")
        count += 1
except StopIteration:
    print("Reached the end of the iterator (StopIteration caught).")

print(f"Total number of consumed characters: {count}")
```

## Exercise 3: Custom iterator class

- Implement class `EvenNumbers(limit)` yielding even numbers up to limit.
- Use `__iter__` and `__next__`.
- Iterate and print values.

```python
# Write your solution here
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        self.current = 0
        return self

    def __next__(self):
        if self.current <= self.limit:
            val = self.current
            self.current += 2
            return val
        else:
            raise StopIteration

print("Even numbers up to 10:")
for num in EvenNumbers(10):
    print(num)
```

## Exercise 4: Iterator state with `next()`

- Create a custom iterator class `Countdown(start)` that counts down to 1.
- Use `iter()` and `next()` to take the first two values manually.
- Then use a `for` loop on the same iterator to get the remaining values.
- Add a short comment explaining why the `for` loop continues where `next()` stopped.

```python
# Write your solution here
class Countdown:
    def __init__(self, start):
        self.start = start
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= 1:
            val = self.current
            self.current -= 1
            return val
        else:
            raise StopIteration

cd = Countdown(5)
cd_iter = iter(cd)

print("Manually fetched values:")
print(next(cd_iter))  # 5
print(next(cd_iter))  # 4

print("Remaining values from for loop:")
for val in cd_iter:
    print(val)

with open("utils_module.py", "w") as f:
    f.write("""def square(n):
    return n * n

def cube(n):
    return n * n * n

def is_even(n):
    return n % 2 == 0
""")

import utils_module

print("Square of 4:", utils_module.square(4))
print("Cube of 3:", utils_module.cube(3))
print("Is 10 even?:", utils_module.is_even(10))

from utils_module import square, is_even

print("Direct import square of 5:", square(5))
print("Direct import is 7 even?:", is_even(7))
```
