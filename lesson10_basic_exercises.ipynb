## Exercise 1: Array basics (using list)

1. Create `cars = ["Ford", "BMW", "Volvo"]`.
2. Print the first and last elements.
3. Print the number of elements.
4. Loop through and print each car.

```python
# Write your solution here
cars = ["Ford", "BMW", "BYD"]

print("First element:", cars[0])
print("Last element:", cars[-1])

print("Number of elements:", len(cars))

print("Cars list:")
for car in cars:
    print(car)
```

## Exercise 2: Array operations

Using `cars`:

```python
# Write your solution here
cars = ["Ford", "BMW", "Volvo"]

cars.append("Honda")

cars.remove("BMW")

removed_car = cars.pop()
print("Removed car:", removed_car)

print("Updated cars list:", cars)
```

## Exercise 3: Built-in iterator

1. Create `nums = [10, 20, 30]`.
2. Create an iterator from `nums` using `iter()`.
3. Use `next()` to print each value one by one.
4. (Optional) Use `try/except` to catch `StopIteration`.

```python
# Write your solution here
nums = [10, 20, 30]

nums_iter = iter(nums)

print("Iterating through values:")
try:
    print(next(nums_iter))
    print(next(nums_iter))
    print(next(nums_iter))
    print(next(nums_iter))  
except StopIteration:
    print("Reached the end of the iterator.")
```

## Exercise 4: Custom iterator class

1. Create a class `CountToFive`.
2. Implement `__iter__()` and `__next__()`.
3. The iterator should return numbers 1 to 5.
4. Use a `for` loop to print all values.

```python
# Write your solution here
class CountToFive:
    def __init__(self):
        self.current = 0
        
    def __iter__(self):
        self.current = 0
        return self
        
    def __next__(self):
        self.current += 1
        if self.current <= 5:
            return self.current
        else:
            raise StopIteration

print("Counting to five using custom iterator:")
for num in CountToFive():
    print(num)
```
