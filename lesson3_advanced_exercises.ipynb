## Exercise 1: Truth table mini-lab

- Create booleans `p = True`, `q = False`.
- Print results for: `p and q`, `p or q`, `not p`, `not q`.
- Add one compound expression with parentheses and print its result.

```python
# Write your solution here
p = True
q = False

print("p and q:", p and q)                  # False
print("p or q:", p or q)                    # True
print("not p:", not p)                      # False
print("not q:", not q)                      # True
print("(p or q) and not q:", (p or q) and not q)  # True
```

## Exercise 2: Comparison chain

- Given `a = 12`, `b = 12`, `c = 7`, print at least 6 comparison results.
- Include equality, inequality, greater/less, and a chained comparison (e.g., `c < a <= b`).

```python
# Write your solution here
a = 12
b = 12
c = 7

print("a == b:", a == b)          # True
print("a != c:", a != c)          # True
print("a > c:", a > c)            # True
print("c < b:", c < b)            # True
print("a < b:", a < b)            # False
print("c < a <= b:", c < a <= b)  # True
```

## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
# Write your solution here
p = True
q = False
r = True

without_parentheses = not p and r or q
with_parentheses = ((not p) and r) or q

print("Without parentheses:", without_parentheses)  # False
print("With parentheses:", with_parentheses)        # False
print("Do they match?", without_parentheses == with_parentheses)  # True
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
# Write your solution here
a = [1, 2, 3]
b = [1, 2, 3]

# == checks whether the lists have equal contents.
# is checks whether both variables refer to the same list object.
print("a == b:", a == b)  # True
print("a is b:", a is b)  # False

c = a  # Now c and a refer to the same list.
print("a == c:", a == c)  # True
print("a is c:", a is c)  # True
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
# Write your solution here
items = ["python", "java", "go", "rust"]
queries = ["python", "javascript", "rust", "ruby"]

for query in queries:
    if query in items:
        print(f"{query}: found")
    elif query not in items:
        print(f"{query}: not found")
```
