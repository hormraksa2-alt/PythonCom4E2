## Exercise 1: Robust numeric parser

- Write a function that parses user-provided string to int or float.
- Raise/handle exceptions for invalid formats.
- Return tuple `(success, value_or_error)` and print examples.

```python
# Write your solution here
def parse_number(s):
    try:
        return (True, int(s))
    except ValueError:
        try:
            return (True, float(s))
        except ValueError as e:
            return (False, str(e))

print("Parse '100':", parse_number("100"))
print("Parse '3.14':", parse_number("3.14"))
print("Parse 'hello':", parse_number("hello"))

def safe_divide(a, b):
    try:
        num1 = float(a)
        num2 = float(b)
        result = num1 / num2
    except TypeError:
        return "Error: Invalid data type provided."
    except ValueError:
        return "Error: Could not convert input to float."
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."
    else:
        return f"Successful division result: {result}"
    finally:
        print("[Log] Division operation attempted.")

print("\n--- Call 1: Normal division ---")
print(safe_divide(20, 4))

print("\n--- Call 2: ZeroDivisionError ---")
print(safe_divide(10, 0))

print("\n--- Call 3: ValueError ---")
print(safe_divide("abc", 5))

print("\n--- Call 4: TypeError ---")
print(safe_divide([1, 2], 2))

data = [15, 25, None, 35, 45, None, 55]

cleaned_list = [x for x in data if x is not None]

if cleaned_list:
    average = sum(cleaned_list) / len(cleaned_list)
else:
    average = 0

print("Cleaned list (valid numbers only):", cleaned_list)
print("Average:", average)
```

## Exercise 2: Multi-exception workflow

- Build a division function that converts both inputs with `float()` and divides, using `try/except/else/finally`.
- Handle `TypeError`, `ValueError`, and `ZeroDivisionError` separately.
- Demonstrate all branches with different calls.

```python
# Write your solution here
def safe_divide(a, b):
    try:
        num1 = float(a)
        num2 = float(b)
        result = num1 / num2
    except TypeError as e:
        print(f"TypeError: Invalid input type ({e})")
        return None
    except ValueError as e:
        print(f"ValueError: Cannot convert input to float ({e})")
        return None
    except ZeroDivisionError:
        print("ZeroDivisionError: Cannot divide by zero")
        return None
    else:
        print(f"Success: {num1} / {num2} = {result}")
        return result
    finally:
        print("Finished division attempt.\n")

safe_divide("10", "2")      
safe_divide("10", "0")     
safe_divide("10", "abc")    
safe_divide([10], "2")      
```

## Exercise 3: None-aware processing

- Given list with values and `None`, compute average of valid numbers only.
- Ignore `None` safely.
- Print cleaned list and average.

```python
# Write your solution here
values = [10, None, 25, None, 15, None, 30]

cleaned = [x for x in values if x is not None and isinstance(x, (int, float)) and not isinstance(x, bool)]

average = sum(cleaned) / len(cleaned) if cleaned else 0

print("Cleaned list:", cleaned)
print("Average:", average)
```

## Exercise 4: Input validation loop

- Prompt repeatedly for integer between 1 and 100 until valid.
- Use exception handling and custom error messages.
- Print accepted value and attempt count.

```python
# Write your solution here
class OutOfRangeError(Exception):
    """Raised when an integer is outside the 1 to 100 range."""
    pass


attempts = 0
accepted_value = None

while True:
    attempts += 1
    try:
        raw = input("Enter an integer between 1 and 100: ").strip()
        value = int(raw)

        if not (1 <= value <= 100):
            raise OutOfRangeError("Number must be between 1 and 100.")

        accepted_value = value
        break
    except ValueError:
        print("Invalid input: please enter a valid whole number (integer).")
    except OutOfRangeError as e:
        print(f"Out of range: {e}")

print(f"Accepted value: {accepted_value}")
print(f"Attempt count: {attempts}")
```

## Exercise 5: Mini CLI app

- Create text menu app (add/subtract/multiply/divide/quit).
- Use input + try/except for invalid choices and values.
- Keep looping until quit command.

```python
# Write your solution here
class InvalidChoiceError(Exception):
    """Raised when the user enters an unsupported menu option."""
    pass


def mini_cli():
    valid_choices = {"add", "subtract", "multiply", "divide", "quit"}

    while True:
        print("\n--- Mini Calculator CLI ---")
        print("Options: add | subtract | multiply | divide | quit")

        try:
            choice = input("Enter choice: ").strip().lower()
            if choice not in valid_choices:
                raise InvalidChoiceError(f"'{choice}' is not a valid menu choice.")

            if choice == "quit":
                print("Exiting CLI app. Goodbye!")
                break

            a = float(input("Enter first number: ").strip())
            b = float(input("Enter second number: ").strip())

            if choice == "add":
                result = a + b
            elif choice == "subtract":
                result = a - b
            elif choice == "multiply":
                result = a * b
            elif choice == "divide":
                result = a / b

        except InvalidChoiceError as e:
            print(f"Choice Error: {e}")
        except ValueError:
            print("Value Error: Please enter valid numeric values.")
        except ZeroDivisionError:
            print("Math Error: Cannot divide by zero.")
        else:
            print(f"Result ({choice}): {result}")


if __name__ == "__main__":
    mini_cli()
```
